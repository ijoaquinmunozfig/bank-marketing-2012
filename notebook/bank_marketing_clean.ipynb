Machine Learning

Train/test split

````python
X = df.drop("y", axis=1)
y = df["y"]
````

````python
X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    X,
    y,
    test_size= 0.25,
    random_state=42,
    stratify=y
    )

print(len(X_train_raw.columns.to_list()), "Training data columns:" ,X_train_raw.columns.to_list())
print(len(X_test_raw.columns.to_list()), "Test data columns: ", X_test_raw.columns.to_list())
````

One-hot encoding (training data)

````python
preprocessor = ColumnTransformer(
    transformers=[(
        "encoder",
        OneHotEncoder(sparse_output=False,
                      drop="first",
                      handle_unknown="ignore"),
        cat_cols)],
        remainder="passthrough"
    )
````

Fit on the training data

````python
preprocessor.fit(X_train_raw)
````

Transform the training data using the fitted preprocessor

````python
X_train_encoded = preprocessor.transform(X_train_raw)
````

Transform the test data using the same preprocessor

````python
X_test_encoded = preprocessor.transform(X_test_raw)
````

Train Models

````python
model_logistic_regresion = LogisticRegression()
model_logistic_regresion.fit(X_train_encoded, y_train)

model_random_forest = RandomForestClassifier()
model_random_forest.fit(X_train_encoded, y_train)
````

Base Models

````python
#Predict
y_pred_LR = model_logistic_regresion.predict(X_test_encoded)
y_pred_RF = model_random_forest.predict(X_test_encoded)
#Probability
y_prob_LR = model_logistic_regresion.predict_proba(X_test_encoded)[:, 1]
y_prob_RF = model_random_forest.predict_proba(X_test_encoded)[:, 1]
````

Base Models Results

````python
print("=== Accuracy ===")
print("Logistic Regression:", accuracy_score(y_test, y_pred_LR))
print("Random Forest:", accuracy_score(y_test, y_pred_RF))

print("\n=== ROC-AUC ===")
print("Logistic Regression:", roc_auc_score(y_test, y_prob_LR))
print("Random Forest:", roc_auc_score(y_test, y_prob_RF))

print("\n=== F1 ===")
print("Logistic Regression:", f1_score(y_test, y_pred_LR))
print("Random Forest:", f1_score(y_test, y_pred_RF))

print("\n=== Recall ===")
print("Logistic Regression:", recall_score(y_test, y_pred_LR))
print("Random Forest:", recall_score(y_test, y_pred_RF))

print("\n=== Precision ===")
print("Logistic Regression:", precision_score(y_test, y_pred_LR))
print("Random Forest:", precision_score(y_test, y_pred_RF))
````

Hiperparameters Optimization

Hiperparameters (Randomized Search CV)

````python
from sklearn.model_selection import RandomizedSearchCV, StratifiedKFold

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

rf_param_dist = {
    "n_estimators": [100, 200, 300, 500],
    "max_depth": [5, 10, 15, 20, None],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 4],
    "class_weight": [None, "balanced"]
}

lr_param_dist = {
    "C": [0.001, 0.01, 0.1, 1, 10, 100],
    "penalty": ["l1", "l2"],
    "solver": ["liblinear"],
    "class_weight": [None, "balanced"]
}

lr_r_search = RandomizedSearchCV(
    estimator=LogisticRegression(
        random_state=42,
        max_iter=1000
    ),
    param_distributions=lr_param_dist,
    n_iter=20,
    cv=cv,
    scoring="f1",
    random_state=42,
    n_jobs=-1
)

rf_r_search = RandomizedSearchCV(
    estimator=RandomForestClassifier(
        random_state=42
    ),
    param_distributions=rf_param_dist,
    n_iter=20,
    cv=cv,
    scoring="f1",
    random_state=42,
    n_jobs=-1
)

lr_r_search.fit(X_train_encoded, y_train)
rf_r_search.fit(X_train_encoded, y_train)

print("Best Logistic Regression Randomized Search params:")
print(lr_r_search.best_params_)

print("\nBest Random Forest Randomized Search params:")
print(rf_r_search.best_params_)
````

````python
lr_r_search_best = LogisticRegression(
    solver="liblinear",
    penalty="l1",
    class_weight="balanced",
    C=100,
    random_state=42,
    max_iter=1000
)

rf_r_search_best = RandomForestClassifier(
    n_estimators=500,
    min_samples_split=2,
    min_samples_leaf=2,
    max_depth=None,
    class_weight="balanced",
    random_state=42
)

lr_r_search_best.fit(X_train_encoded, y_train)
rf_r_search_best.fit(X_train_encoded, y_train)
````

````python
#Predict
y_pred_LR_search_best = lr_r_search_best.predict(X_test_encoded)
y_pred_RF_search_best = rf_r_search_best.predict(X_test_encoded)
#Probability
y_prob_LR_search_best = lr_r_search_best.predict_proba(X_test_encoded)[:, 1]
y_prob_RF_search_best = rf_r_search_best.predict_proba(X_test_encoded)[:, 1]
````

Hiperparameters (Optuna)

````python
import optuna
from sklearn.model_selection import cross_val_score

def optuna_objective_lr(trial):
    model = LogisticRegression(
        C=trial.suggest_float(
            "C",
            0.001,
            100,
            log=True
        ),
        penalty=trial.suggest_categorical(
            "penalty",
            ["l1", "l2"]
        ),
        class_weight=trial.suggest_categorical(
            "class_weight",
            [None, "balanced"]
        ),
        solver="liblinear",
        max_iter=1000,
        random_state=42
    )

    optuna_score_lr = cross_val_score(
        model,
        X_train_encoded,
        y_train,
        cv=5,
        scoring="f1"
    ).mean()

    return optuna_score_lr
````

````python
optuna_study_lr = optuna.create_study(
    direction="maximize"
)

optuna_study_lr.optimize(
    optuna_objective_lr,
    n_trials=10
)
````

````python
optuna_study_lr.best_params
````

````python
lr_optuna_best = LogisticRegression(
    **optuna_study_lr.best_params,
    solver="liblinear",
    random_state=42,
    max_iter=1000
)

lr_optuna_best.fit(X_train_encoded,y_train)
````

````python
#Predict
y_pred_LR_optuna_best = lr_optuna_best.predict(X_test_encoded)
#Probability
y_prob_LR_optuna_best = lr_optuna_best.predict_proba(X_test_encoded)[:, 1]
````

````python
def optuna_objective_rf(trial):
    model = RandomForestClassifier(
            n_estimators=trial.suggest_int(
                "n_estimators", 100, 500
            ),
            max_depth=trial.suggest_int(
                "max_depth", 5, 20
            ),
            min_samples_split=trial.suggest_int(
                "min_samples_split", 2, 10
            ),
            min_samples_leaf=trial.suggest_int(
                "min_samples_leaf", 1, 4
            ),
            random_state=42
        )

    optuna_score_rf = cross_val_score(
            model,
            X_train_encoded,
            y_train,
            cv=5,
            scoring="f1"
        ).mean()

    return optuna_score_rf
````

````python
optuna_study_rf = optuna.create_study(
    direction="maximize"
)

optuna_study_rf.optimize(
    optuna_objective_rf,
    n_trials=50
)
````

````python
%%writefile -a app/streamlit_app.py


@st.cache_data(show_spinner="Loading data from the UCI repository...")
def load_data():
    from ucimlrepo import fetch_ucirepo

    bank = fetch_ucirepo(id=222)
    data = pd.concat([bank.data.features, bank.data.targets], axis=1)
    data = data.assign(contact_bool=(data["contact"] == "cellular"))
    data = data.drop(columns=["contact", "duration"])
    for col in CAT_COLS:
        data[col] = data[col].fillna("missing")
    for col in ["default", "housing", "loan", "y"]:
        data[col] = data[col].eq("yes")
    return data


model = load_model()
meta = load_metadata()
FEATURES = meta["features"]
OPTIONS = meta["categorical_options"]
RANGES = meta["numeric_ranges"]
````

````python
%%writefile -a app/streamlit_app.py

st.title("Term Deposit Propensity Model")
st.caption(
    "Educational portfolio project built on historical campaign data from a Portuguese bank. "
    "It is not a tool for real customer decisions."
)

tab_pred, tab_model, tab_data = st.tabs(["Prediction", "Model performance", "Data exploration"])

with tab_pred:
    proba = float(model.predict_proba(client)[0, 1])
    base_rate = meta["dataset"]["train_positive_rate"]

    col1, col2, col3 = st.columns(3)
    col1.metric("Probability of subscription", f"{proba:.1%}")
    col2.metric("Average in training data", f"{base_rate:.1%}")
    col3.metric("Relative to average", f"{proba / base_rate:.1f}x")

    if proba >= threshold:
        st.success(f"Above the {threshold:.0%} threshold: this profile would be prioritized for contact.")
    else:
        st.info(f"Below the {threshold:.0%} threshold: this profile would not be prioritized.")

    gauge = go.Figure(go.Indicator(
        mode="gauge+number",
        value=proba * 100,
        number={"suffix": "%"},
        gauge={
            "axis": {"range": [0, 100]},
            "bar": {"color": NAVY},
            "threshold": {"line": {"color": "#EF4444", "width": 3}, "value": threshold * 100},
        },
    ))
    gauge.update_layout(height=260, margin=dict(l=20, r=20, t=20, b=10))
    st.plotly_chart(gauge)

    with st.expander("Input features sent to the model"):
        st.dataframe(friendly_frame(client), hide_index=True)
````

Model performance

````python
%%writefile -a app/streamlit_app.py

with tab_model:
    scores = load_scores()
    y_true = scores["y_true"]
    y_prob = scores["y_prob"]
    y_hat = (y_prob >= threshold).astype(int)

    st.subheader("Performance on the held-out test set")
    st.caption(
        f"{len(scores):,} customers not used for training. "
        "The metrics change with the decision threshold in the sidebar."
    )

    c1, c2, c3, c4 = st.columns(4)
    c1.metric("Precision", f"{precision_score(y_true, y_hat, zero_division=0):.1%}")
    c2.metric("Recall", f"{recall_score(y_true, y_hat, zero_division=0):.1%}")
    c3.metric("F1 score", f"{f1_score(y_true, y_hat, zero_division=0):.1%}")
    c4.metric("ROC-AUC", f"{roc_auc_score(y_true, y_prob):.3f}")

    left, right = st.columns(2)

    tn, fp, fn, tp = confusion_matrix(y_true, y_hat, labels=[0, 1]).ravel()
    cm = px.imshow(
        [[tn, fp], [fn, tp]],
        x=["Predicted: no", "Predicted: yes"],
        y=["Actual: no", "Actual: yes"],
        text_auto=True,
        color_continuous_scale="Blues",
        title="Confusion matrix",
    )
    cm.update_coloraxes(showscale=False)
    left.plotly_chart(cm)

    grid = [i / 100 for i in range(5, 96, 5)]
    curve = pd.DataFrame({
        "threshold": grid,
        "Precision": [precision_score(y_true, (y_prob >= t).astype(int), zero_division=0) for t in grid],
        "Recall": [recall_score(y_true, (y_prob >= t).astype(int), zero_division=0) for t in grid],
    })
    trade = px.line(
    curve, x="threshold", y=["Precision", "Recall"], markers=True,
    title="Precision and recall by threshold",
    labels={"threshold": "Decision threshold", "value": "Score", "variable": "Metric"})
    trade.add_vline(x=threshold, line_dash="dash")
    right.plotly_chart(trade)

    importance = load_importance().head(10).assign(feature=lambda d: d["feature"].map(feature_label))
    fig_imp = px.bar(
    importance, x="importance", y="feature", orientation="h",
    title="Top 10 variables by importance",
    labels={"importance": "Importance", "feature": "Variable"})
    fig_imp.update_traces(marker_color=NAVY)
    fig_imp.update_layout(yaxis={"categoryorder": "total ascending"})
    st.plotly_chart(fig_imp)
    st.caption(
        "Impurity-based importance shows what the forest uses to split the data, not a causal effect. "
        "Continuous variables such as balance and age tend to rank higher."
    )

    with st.expander("About the model"):
        st.write(f"Model: {meta['model_name']}")
        st.write(f"Trained on: {meta['trained_on']}")
        st.write(f"Excluded to avoid data leakage: {', '.join(meta['excluded_features'])}")
        st.write(f"Data: {meta['dataset']['source']} ({meta['dataset']['license']})")
````
